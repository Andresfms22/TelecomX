 Informe Final — Análisis de Evasión de Clientes (Churn) en Telecom X

1. Introducción
La evasión de clientes (Churn) es uno de los principales desafíos para las empresas de telecomunicaciones. Comprender por qué los clientes abandonan permite diseñar estrategias efectivas de retención, optimizar la experiencia del usuario y reducir pérdidas económicas.
El objetivo de este análisis es:
- Explorar los datos proporcionados por la API de Telecom X.
- Identificar patrones asociados al abandono de clientes.
- Determinar las variables más influyentes en el churn.
- Proponer recomendaciones estratégicas basadas en evidencia.
Este informe resume el proceso completo: desde la carga y limpieza de datos hasta el análisis exploratorio y las conclusiones finales.

2. Limpieza y Tratamiento de Datos
 Carga de datos
Los datos fueron importados directamente desde la API proporcionada por el desafío y convertidos en un DataFrame de Pandas para su análisis.
 Revisión inicial
Se identificaron 21 columnas con información demográfica, contractual, de servicios y facturación.
 Tratamiento aplicado
- Eliminación de duplicados para asegurar integridad.
- Conversión de tipos de datos, especialmente:
- account.Charges.Total → float
- Churn → variable binaria (1 = abandono, 0 = permanencia)
- Eliminación de registros con Churn nulo (224 filas).
- Estandarización de categorías, reemplazando valores como "No internet service" por "No".
- Verificación final para asegurar consistencia y ausencia de nulos.
El dataset final quedó con 7043 registros limpios listos para el análisis.

3. Análisis Exploratorio de Datos (EDA)
El EDA se realizó en tres niveles: numérico, categórico y combinado.

3.1 Análisis Numérico
 Tenure (meses como cliente)
- Los clientes que abandonan tienen tenure significativamente menor.
- La evasión ocurre principalmente en los primeros meses.
 Monthly Charges (cargo mensual)
- Los clientes con cargos mensuales más altos muestran mayor churn.
- Indica sensibilidad al precio.
 Total Charges (total gastado)
- Los clientes que abandonan tienen total gastado bajo, coherente con su poco tiempo en la empresa.

3.2 Análisis Categórico
 Tipo de contrato
- Month-to-month presenta la mayor tasa de churn.
- Contratos de 1 y 2 años muestran churn muy bajo.
 Tipo de servicio de Internet
- Fiber optic tiene el churn más alto.
- DSL y No Internet presentan menor abandono.
 Servicios adicionales
Clientes sin:
- OnlineSecurity
- TechSupport
- DeviceProtection
presentan mayor churn.
 Método de pago
- Electronic check está asociado a la mayor tasa de abandono.
 SeniorCitizen
- Los adultos mayores muestran mayor churn.
 Género, Partner, Dependents
- No presentan influencia significativa.

3.3 Análisis Combinado
El cruce de variables reveló patrones más profundos:
- Tenure + Contract:
Clientes nuevos con contrato mensual son el grupo con mayor churn.
- Monthly Charges + InternetService:
Los usuarios de fibra óptica pagan más y abandonan más.
- PaymentMethod + MonthlyCharges:
Electronic check + cargos altos = mayor probabilidad de abandono.
- OnlineSecurity + Tenure:
Los clientes con servicios de seguridad permanecen más tiempo.

4. Conclusiones e Insights
Los factores más influyentes en la evasión de clientes son:
 1. Tipo de contrato
El predictor más fuerte del churn.
Los contratos mensuales facilitan la salida del cliente.
 2. Tenure
La evasión ocurre principalmente en los primeros meses.
La experiencia inicial es crítica.
 3. Cargos mensuales altos
Los clientes con costos elevados abandonan más.
 4. Tipo de servicio de Internet
La fibra óptica presenta mayor churn, posiblemente por problemas de calidad o expectativas.
 5. Servicios adicionales
La ausencia de OnlineSecurity y TechSupport aumenta la probabilidad de abandono.
 6. Método de pago
Electronic check está asociado a mayor churn.
 7. SeniorCitizen
Los adultos mayores requieren mayor soporte.

5. Recomendaciones Estratégicas
Basadas en los hallazgos, se sugieren las siguientes acciones:

 1. Fortalecer la experiencia inicial del cliente
- Programas de bienvenida.
- Seguimiento proactivo durante los primeros 3 meses.
- Soporte prioritario para nuevos clientes.

 2. Incentivar contratos de mayor duración
- Descuentos por permanencia.
- Beneficios exclusivos para contratos anuales.
- Campañas de retención dirigidas a clientes con contrato mensual.

 3. Revisar la oferta de fibra óptica
- Evaluar calidad del servicio.
- Mejorar estabilidad y velocidad.
- Ajustar precios o agregar valor adicional.

 4. Promover servicios adicionales
- Paquetes con OnlineSecurity y TechSupport.
- Ofertas combinadas para aumentar retención.
 5. Optimizar la experiencia de pago
- Incentivar métodos automáticos (tarjeta, transferencia).
- Reducir fricción en Electronic check.
 6. Atención especial a adultos mayores
- Soporte personalizado.
- Material educativo simplificado.
- Canales de atención preferenciales.

Cierre
